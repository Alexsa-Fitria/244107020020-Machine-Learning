### **Langkah 0: Persiapan (Impor Library & Load Data)**

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load data
df = pd.read_csv('wbc.csv')
df.head()
```

### Tugas 1: Pisahkan Variabel yang Dapat dan Tidak Dapat Digunakan

```python
df.info()
```

```python
# Kolom yang tidak dapat digunakan (dibuang)
drop_cols = ['id', 'Unnamed: 32']

# Variabel yang dapat digunakan (fitur)
X = df.drop(columns=drop_cols + ['diagnosis'])

# Variabel target (yang mau diprediksi)
y = df['diagnosis']

print("Fitur yang digunakan:", X.columns.tolist())
```

### Tugas 2: Encoding pada Kolom "diagnosis"

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

# Cek hasilnya
print(df['diagnosis'].value_counts())
df.head()
```

### Tugas 3: Standardisasi Semua Kolom Numerik

```python
# Ambil semua kolom numerik (kecuali diagnosis karena itu target)
numeric_cols = X.select_dtypes(include=['float64', 'int64']).columns

# Lakukan standardisasi
scaler = StandardScaler()
X[numeric_cols] = scaler.fit_transform(X[numeric_cols])

# Cek hasilnya
X.head()
```

### Verifikasi Akhir

```python
print("Shape X:", X.shape)
print("Shape y:", y.shape)
print("\nCek missing value:")
print(X.isnull().sum().sum())
print("\nStatistik X setelah standardisasi:")
print(X.describe().loc[['mean', 'std']])
```
